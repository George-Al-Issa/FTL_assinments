```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("FAOSTAT_data_forest_emissionssami.csv")

print(df.head())
print(df.info())
```

```python
print("\nElements:")
print(df["Element"].unique())

print("\nItems:")
print(df["Item"].unique())

print("\nYears:")
print(df["Year"].unique())
```

```python
df = df[(df["Year"] >= 2000) & (df["Year"] <= 2024)]

print(df["Year"].unique())
print("Number of years:", df["Year"].nunique())
```

```python
forest_emissions = df[
    (df["Element"] == "Net emissions") &
    (df["Item"] == "Forestland")
]

print("\nForestland Net Emissions:")
print(forest_emissions[["Year", "Value", "Unit"]])
```

```python
average = forest_emissions["Value"].mean()
max_value = forest_emissions["Value"].max()
min_value = forest_emissions["Value"].min()
max_year = forest_emissions.loc[forest_emissions["Value"].idxmax(), "Year"]
min_year = forest_emissions.loc[forest_emissions["Value"].idxmin(), "Year"]

print("\n--- Results ---")
print("Average net emissions:", average, "kt")
print("Maximum value:", max_value, "kt in", max_year)
print("Minimum value:", min_value, "kt in", min_year)
```

```python
value_2000 = forest_emissions.loc[forest_emissions["Year"] == 2000, "Value"].iloc[0]
value_2024 = forest_emissions.loc[forest_emissions["Year"] == 2024, "Value"].iloc[0]
absolute_change = value_2024 - value_2000
percentage_change = (absolute_change / abs(value_2000)) * 100

print("\n--- Change Analysis ---")
print("Net emissions in 2000:", value_2000, "kt")
print("Net emissions in 2024:", value_2024, "kt")
print("Absolute change:", absolute_change, "kt")
print("Percentage change:", percentage_change, "%")
```

```python
years = forest_emissions["Year"]
values = forest_emissions["Value"]
slope = values.cov(years) / years.var()

print("\n--- Trend Analysis ---")
print("Annual trend:", slope, "kt/year")
```

```python
plt.figure(figsize=(10, 5))
plt.plot(forest_emissions["Year"], forest_emissions["Value"], marker="o")
plt.xlabel("Year")
plt.ylabel("Net emissions (kt)")
plt.title("Forestland Net Emissions in Syria (2000–2024)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
early_period = forest_emissions[forest_emissions["Year"].between(2000, 2004)]["Value"].mean()
recent_period = forest_emissions[forest_emissions["Year"].between(2020, 2024)]["Value"].mean()
period_change = recent_period - early_period

print("\n--- Period Comparison ---")
print("Average net emissions (2000-2004):", early_period, "kt")
print("Average net emissions (2020-2024):", recent_period, "kt")
print("Difference between periods:", period_change, "kt")
```
