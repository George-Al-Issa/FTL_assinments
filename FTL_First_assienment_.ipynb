```python
class ClimateStation :# class creation with the three methods

  def __init__(self, city, Temperture,Humidity,rainfall):
    self.cityname = city
    self.temp = Temperture
    self.humidity = Humidity
    self.rain = rainfall

  def summary(self):
    return print("city name is",self.cityname,"\n temperture",self.temp,"\n humidity is",self.humidity,"\n rain chance",self.rain)

  def calculate_risk(self):

    if self.temp in range(40,46):
      return 'Extreme'
    elif self.temp in range(35,39) or self.rain<5:
      return 'High'
    elif self.temp in range(30,34) or self.rain<15:
      return 'Moderate'
    else:
      return 'Low'

  def update_temperuture(self,new_temp):
    self.temp=input("please entre the updated temperture")
    return self.temp


```

```python
class SmartClimateStation(ClimateStation):
  def __init__(self, sensor_status):
   self.sessor_status=sensor_status

  def check_sensor(self):
    if self.sessor_status<0:
      return 'low'
    elif self.sessor_status>0 and self.sessor_status<10:
      return 'moderate'
    elif self.sessor_status>10 and self.sessor_status<20:
      return 'high'
    else:
      return 'extreme'

  def recommendation(self):# recomend a response based on danger level
      check_sensor=self.check_sensor()
      if check_sensor == 'low':
        send_SMS=False
        alert=False
        print('all is good, no action needed')
      elif check_sensor == 'moderate':
        send_SMS=True # for future automation
        alert=False

        print('send normal sms')
      elif check_sensor == 'high':
        send_SMS=True
        alert=False

        print('send warning sms of high caution')
      else:
        send_SMS=True

        alert=True# to turn on the alaram
        print('send warning sms of extreme priorty and activate the alert')
      return send_SMS, alert

```
